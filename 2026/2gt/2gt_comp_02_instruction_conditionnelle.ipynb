Dans ce carnet, certains programmes sont là pour être **lus** : ils n'ont pas de crochets `[ ]` à leur gauche et ne s'exécutent pas. Les autres sont des cellules de code : clique dedans, puis `Maj` + `Entrée`.

## Cours

À lire avant de commencer.

Jusqu'ici, un programme exécutait toutes ses lignes, dans l'ordre, une fois chacune. L'**instruction conditionnelle** brise cette règle : elle laisse le programme **choisir** ce qu'il exécute.

````python
cote = 12
if cote > 10:
    print("grand")
print("fin")
````

La ligne `if` pose une **condition** et se termine par deux-points. Les lignes qui suivent sont décalées vers la droite de quatre espaces : c'est cette **indentation**, et elle seule, qui dit ce qui dépend du test. Si la condition est vraie, le bloc décalé s'exécute ; si elle est fausse, il est **sauté en entier**. Dans les deux cas, le programme reprend à la première ligne non décalée.

`else:` ouvre le bloc réservé au cas où la condition est fausse. `elif` — lire « sinon si » — ajoute un test supplémentaire.

````python
angle = 40
if angle > 90:
    print("obtus")
elif angle == 90:
    print("droit")
else:
    print("aigu")
````

Les tests sont examinés **dans l'ordre où ils sont écrits**. Le premier qui réussit gagne, et les suivants ne sont même pas lus. Deux programmes qui posent les mêmes tests dans un ordre différent ne classent donc pas de la même façon.

Une condition se construit avec un **comparateur**. Tu en connais deux ; en voici six.

| Écriture | Sens |
|---|---|
| `a < b` | a est strictement inférieur à b |
| `a > b` | a est strictement supérieur à b |
| `a <= b` | a est inférieur ou égal à b |
| `a >= b` | a est supérieur ou égal à b |
| `a == b` | a est égal à b |
| `a != b` | a est différent de b |

Deux conditions se relient par `and` — il faut alors que les deux soient vraies — ou par `or`, qui se contente d'une seule des deux. Enfin, `print("dedans")` affiche un message : le texte entre guillemets est une chaîne de caractères.

**Le piège de cette notion.** Le signe `=` **range** une valeur dans une variable ; le signe `==` **compare** deux valeurs. Les confondre est l'erreur la plus fréquente de l'année, et elle a deux visages.

Le premier est bruyant. `if cote = 10:` ne veut rien dire : Python refuse de démarrer et signale une `SyntaxError`. On la corrige en trois secondes.

Le second est muet. La ligne `cote == 10`, écrite seule, est acceptée : Python compare, obtient vrai ou faux, puis jette cette réponse que personne ne recueille. La variable n'a pas bougé, aucun message n'apparaît, et le programme continue avec l'ancienne valeur. C'est ce second visage que tu rencontreras.

````python
print("Note nécessaire au dernier DS :", note)
````

```python
else:
    print("Objectif atteignable.")
```

Le programme affiche **21.5**, puis **Objectif impossible à atteindre.**

Le seuil s'écrit `> 20` parce qu'un 20 est une note qui existe : elle est difficile, elle n'est pas impossible. Avec `>= 20`, un élève à qui il faudrait précisément 20 s'entendrait répondre que son objectif est hors d'atteinte, alors qu'il ne l'est pas.

- **jeu 3** : seul D a bougé d'une unité par rapport au jeu de départ, et cela suffit. Le déplacement de D vers C vaut maintenant (4 ; 2) contre (5 ; 2) pour A vers B : ce n'est même plus un parallélogramme, alors que le dessin a à peine changé.
- **jeu 4** : ce n'est pas un parallélogramme non plus — (3 ; 4) contre (6 ; 8) —, et pourtant AB² = BC² = 25. Un programme qui testerait le losange **en premier** répondrait « losange » à propos d'un quadrilatère dont les quatre côtés mesurent 5, 5, 10 et 8. Deux côtés consécutifs égaux ne font pas un losange : il y faut d'abord un parallélogramme.

C'est tout le rôle du `elif`. Il n'est examiné que si le test précédent a échoué : la question « est-ce un losange ? » n'est donc posée qu'à un quadrilatère dont on sait déjà qu'il est un parallélogramme. L'ordre des tests fait ici un travail mathématique, il ne range pas seulement des lignes.

Attention aussi au connecteur. En écrivant `and` au lieu de `or` dans le premier test, on n'écarterait que les quadrilatères dont les **deux** coordonnées du déplacement diffèrent. Le jeu 3, dont seule l'abscisse diffère, serait alors annoncé comme un parallélogramme, sans le moindre message d'erreur.

*Prolongement de l'exercice `p2`, adapté du manuel de 2de, page 167, exercice 72 : la fonction du manuel testait un sommet désigné, celle-ci les cherche tous les trois.*

À l'exercice `p2`, l'énoncé te disait où chercher : le programme testait le sommet B, et lui seul. Il pouvait donc répondre « non » à propos d'un triangle rectangle ailleurs, sans jamais s'en apercevoir.

Écris un programme qui pose les six coordonnées de A, B et C en tête, affiche AB², BC² et AC², puis affiche **une seule** de ces quatre phrases :

Il doit répondre juste pour les trois triangles suivants, en ne changeant que ses six premières lignes.

| | A | B | C |
|---|---|---|---|

| triangle 2 | (1 ; 2) | (5 ; 3) | (0 ; 6) |
| triangle 3 | (−1 ; 3) | (6 ; 7) | (2 ; 1) |

**Contraintes** : après les six premières lignes, aucune coordonnée ne réapparaît dans ton programme, et il n'y a de racine carrée nulle part.

Sur ton cahier, pour finir : l'égalité AC² = AB² + BC² place l'angle droit en B. Explique pourquoi c'est bien B, et non A ni C.

<details>
<summary><strong>Un coup de pouce</strong></summary>

Trois sommets, donc trois égalités possibles. Écris-les toutes les trois sur ton cahier avant de programmer : dans chacune, l'un des trois carrés est la somme des deux autres.

Pour savoir quel sommet porte l'angle droit dans l'égalité `AC2 == AB2 + BC2`, regarde quels côtés y interviennent. [AB] et [BC] se rencontrent en B ; le côté qui reste, [AC], est l'hypoténuse, et l'hypoténuse est en face de l'angle droit.

Une seule phrase doit sortir, quoi qu'il arrive. C'est ce que garantit une suite de `elif` : dès qu'un test réussit, les suivants ne sont même pas lus.

</details>

<details>
<summary><strong>Correction</strong></summary>

```python
xA = -1
yA = 3
xB = 6
yB = 7
xC = 2
yC = 1
AB2 = (xA - xB) * (xA - xB) + (yA - yB) * (yA - yB)
BC2 = (xB - xC) * (xB - xC) + (yB - yC) * (yB - yC)
AC2 = (xA - xC) * (xA - xC) + (yA - yC) * (yA - yC)
print("AB² =", AB2, "  BC² =", BC2, "  AC² =", AC2)
if AC2 == AB2 + BC2:
    print("Le triangle est rectangle en B.")
elif BC2 == AB2 + AC2:
    print("Le triangle est rectangle en A.")
elif AB2 == AC2 + BC2:
    print("Le triangle est rectangle en C.")
else:
    print("Le triangle n'est rectangle en aucun de ses sommets.")
```

Sur le troisième triangle, il affiche :

    AB² = 65   BC² = 52   AC² = 13
    Le triangle est rectangle en C.

Les trois verdicts :

- **triangle 1** : AB² = 24,25, BC² = 29, AC² = 55,25. La somme AB² + BC² vaut 53,25 contre 55,25 : il s'en faut de 2. Le triangle n'est rectangle nulle part — et l'angle en B mesure pourtant 92,16°, un écart de deux degrés qu'aucun œil ne détecte. C'est ce que `p2` montrait déjà ; ce que cet exercice ajoute, c'est qu'il n'est pas non plus rectangle ailleurs, ce que `p2` ne pouvait pas savoir.
- **triangle 2** : AB² = 17, BC² = 34, AC² = 17. On a BC² = AB² + AC², donc l'angle droit est en A. Et AB² = AC², donc le triangle est isocèle en A par-dessus le marché.
- **triangle 3** : AB² = 65, BC² = 52, AC² = 13, et 65 = 13 + 52. L'angle droit est en C.

**Le sommet de l'angle droit.** Dans l'égalité AC² = AB² + BC², le plus grand carré est celui de [AC] : ce côté est l'hypoténuse, et l'angle droit lui fait face. Or, des trois sommets, un seul n'est pas une extrémité de [AC] : c'est B. Voilà pourquoi cette égalité-là désigne B, et pourquoi il faut trois tests et non un. Le programme de `p2`, qui ne connaissait que celui-ci, aurait répondu « pas rectangle » aux triangles 2 et 3, qui le sont pourtant tous les deux.

**Les deux contraintes** travaillent ensemble. Les trois carrés sont calculés une fois, nommés, et les tests ne parlent plus que de ces trois noms : changer de triangle demande alors six modifications, toutes groupées en tête, et pas une de plus. Quant aux racines carrées, elles ne serviraient à rien et coûteraient l'exactitude. Ici, 65, 52 et 13 sont des entiers, et l'égalité 65 = 13 + 52 est vraie sans discussion. En passant par les longueurs — `AB = 65 ** 0.5`, puis `AB * AB` —, Python ne retrouve même plus 65, mais 64.99999999999999. Le verdict tiendrait encore, par chance ; on aurait simplement construit un test dont plus rien ne garantit la réponse.
