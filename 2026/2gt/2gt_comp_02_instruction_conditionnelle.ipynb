## Cours

À lire avant de commencer.

Jusqu'ici, un programme exécutait toutes ses lignes, dans l'ordre, une fois chacune. L'**instruction conditionnelle** brise cette règle : elle laisse le programme **choisir** ce qu'il exécute.

````python
cote = 12
if cote > 10:
    print("grand")
print("fin")
````

La ligne `if` pose une **condition** et se termine par deux-points. Les lignes qui suivent sont décalées vers la droite de quatre espaces : c'est cette **indentation**, et elle seule, qui dit ce qui dépend du test. Si la condition est vraie, le bloc décalé s'exécute ; si elle est fausse, il est **sauté en entier**. Dans les deux cas, le programme reprend à la première ligne non décalée.

`else:` ouvre le bloc réservé au cas où la condition est fausse. `elif` — lire « sinon si » — ajoute un test supplémentaire.

````python
angle = 40
if angle > 90:
    print("obtus")
elif angle == 90:
    print("droit")
else:
    print("aigu")
````

Les tests sont examinés **dans l'ordre où ils sont écrits**. Le premier qui réussit gagne, et les suivants ne sont même pas lus. Deux programmes qui posent les mêmes tests dans un ordre différent ne classent donc pas de la même façon.

Une condition se construit avec un **comparateur**. Tu en connais deux ; en voici six.

| Écriture | Sens |
|---|---|
| `a < b` | a est strictement inférieur à b |
| `a > b` | a est strictement supérieur à b |
| `a <= b` | a est inférieur ou égal à b |
| `a >= b` | a est supérieur ou égal à b |
| `a == b` | a est égal à b |
| `a != b` | a est différent de b |

Deux conditions se relient par `and` — il faut alors que les deux soient vraies — ou par `or`, qui se contente d'une seule des deux. Enfin, `print("dedans")` affiche un message : le texte entre guillemets est une chaîne de caractères.

**Le piège de cette notion.** Le signe `=` **range** une valeur dans une variable ; le signe `==` **compare** deux valeurs. Les confondre est l'erreur la plus fréquente de l'année, et elle a deux visages.

Le premier est bruyant. `if cote = 10:` ne veut rien dire : Python refuse de démarrer et signale une `SyntaxError`. On la corrige en trois secondes.

Le second est muet. La ligne `cote == 10`, écrite seule, est acceptée : Python compare, obtient vrai ou faux, puis jette cette réponse que personne ne recueille. La variable n'a pas bougé, aucun message n'apparaît, et le programme continue avec l'ancienne valeur. C'est ce second visage que tu rencontreras.

```python
from turtle import *
reset()

n = 4

if n == 3:
    angle = 120
else:
    angle = 90

forward(100)
left(angle)
forward(100)
left(angle)
forward(100)
left(angle)

if n == 4:
    forward(100)
    left(angle)

done()
```

Les deux figures ont des côtés de même longueur : c'est le pivot et le nombre de côtés qui changent. Le premier `if` range l'angle du pivot dans une variable, exactement comme le faisait l'exercice `m1` avec une longueur — une condition sert au moins aussi souvent à préparer un calcul qu'à afficher un message.

<details>
<summary><strong>Un coup de pouce</strong></summary>

Personne ne sait à l'avance laquelle des trois longueurs est la plus grande : ton programme doit le découvrir. Pars du principe que c'est la première, range-la dans une variable, puis corrige cette variable si la deuxième la dépasse, puis si la troisième la dépasse.

Il te faut ensuite la somme des deux autres, sans savoir lesquelles ce sont. Que vaut la somme des trois longueurs, diminuée de la plus grande ?

</details>

<details>
<summary><strong>Correction</strong></summary>

```python
a = 7
b = 3
c = 4
plus_grand = a
if b > plus_grand:
    plus_grand = b
if c > plus_grand:
    plus_grand = c
somme_autres = a + b + c - plus_grand
print(plus_grand)
if plus_grand < somme_autres:
    print("possible")
else:
    print("impossible")
```

Les deux premiers `if` n'ont pas de `else`, et c'est voulu : quand la condition est fausse, il n'y a rien à faire, la variable garde ce qu'elle avait. Deux tests successifs suffisent à départager trois nombres, parce que chaque test compare un candidat au meilleur trouvé jusque-là.

L'astuce de la dernière ligne évite d'avoir à savoir *lesquelles* sont les deux autres longueurs : leur somme est celle des trois, moins la plus grande.

Avec (6 ; 5 ; 4), le programme affiche 6 puis « possible ». Avec (2 ; 9 ; 4), il affiche 9 puis « impossible » : 9 dépasse 2 + 4. Avec (7 ; 3 ; 4), il affiche 7 puis « impossible », et ce cas mérite un dessin : 3 + 4 vaut exactement 7, les trois points sont alignés et le triangle est aplati. C'est pour lui que la condition exige une inégalité **stricte**.
