## Cours

À lire avant de commencer.

Un **programme** est une suite d'instructions. Elles sont exécutées **dans l'ordre**, de haut en bas, une seule fois chacune. Échanger deux lignes, c'est changer le programme.

Une **variable** est un nom qui retient une valeur. On la crée par une **affectation**. En langage naturel, l'affectation s'écrit avec une flèche :

    cote ← 5

qui se lit « `cote` reçoit 5 ». En Python, cette flèche s'écrit avec le signe égal.

````python
cote = 5
print(cote)
````

**Ce signe n'est pas une égalité mathématique.** Il ne compare rien : il range une valeur dans une variable. La ligne se lit de droite à gauche — on calcule ce qui est à droite, on le range dans le nom qui est à gauche. C'est pourquoi la ligne `cote = cote + 3` n'a aucun sens en mathématiques, alors qu'elle en a un très précis en Python : prendre la valeur actuelle de `cote`, lui ajouter 3, ranger le total dans `cote`.

**Le piège de cette notion.** Une variable garde la valeur qu'elle avait **au moment où sa ligne a été lue**. Si elle change plus tard, ce qui a déjà été calculé à partir d'elle ne change pas pour autant. Ce piège se voit parfois du premier coup d'œil ; il sait aussi rester totalement silencieux. Tu le rencontreras sous ses deux visages.

Toute valeur a un **type**. Il y en a quatre à connaître.

| Type | Nom en Python | Exemples |
|---|---|---|
| entier | `int` | `4`, `-17`, `0` |
| flottant, c'est-à-dire nombre à virgule | `float` | `3.5`, `2.0` |
| chaîne de caractères | `str` | `"carré"`, `"7"` |
| booléen | `bool` | `True`, `False` |

Trois faits à retenir. La division `/` donne **toujours** un flottant, même quand le quotient tombe juste. Deux chaînes s'ajoutent en se collant l'une à l'autre : `"7" + "7"` donne `"77"`. Une comparaison comme `cote > 10` ou `cote == 5` ne donne pas un nombre mais un booléen — attention : `==` compare, tandis que `=` range. L'instruction `type(cote)` affiche le type d'une variable.

````python
print(7 / 2)
print(type(7 / 2))
````

**La tortue.** Quelques instructions suffisent à dessiner. `forward(60)` avance de 60 pas en laissant un trait. `left(90)` et `right(90)` font pivoter la tortue sur place d'un quart de tour, vers sa gauche ou vers sa droite, sans rien tracer. `reset()` repart d'un écran vide et `done()` termine le dessin. La tortue démarre au centre, tournée vers la droite.

### ◆ Exercice e1 — ÉCRIRE

Difficulté ●●●○○

*Adapté du TP « Calculer le périmètre et l'aire d'un triangle avec la formule de Héron », Planète Maths, académie de Grenoble : la fonction a été retirée et la comparaison avec le carré ajoutée.*

La clôture ne pose aucune difficulté. L'aire, si : on ne connaît aucune hauteur de ce triangle, donc la formule « base × hauteur / 2 » est inutilisable ici. Héron d'Alexandrie a laissé une formule qui n'utilise que les trois côtés. En appelant `a`, `b` et `c` les trois côtés, et `p` le demi-périmètre, c'est-à-dire la moitié du périmètre :

Deux choses à savoir pour l'écrire en Python. La racine carrée d'un nombre s'obtient en l'élevant à la puissance 0,5, ce qui s'écrit `(...) ** 0.5`. Et `round(x, 1)` donne la valeur de `x` arrondie au dixième.

````python
# Écris ton code ici
````

<details>
<summary><strong>Un coup de pouce</strong></summary>

Le demi-périmètre sert quatre fois dans la formule. Donne-lui un nom avant de t'en servir, sinon tu vas le recopier quatre fois — et le corriger quatre fois le jour où un côté changera.

Le carré a la même clôture que le triangle. Combien mesure alors chacun de ses quatre côtés ?

Pour la question du cahier : ce qu'on perd, c'est la différence entre les deux aires. Un pourcentage se calcule toujours par rapport à quelque chose — ici, par rapport à l'aire du carré.

</details>

<details>
<summary><strong>Correction</strong></summary>

```python
a = 37
b = 53
c = 68
p = (a + b + c) / 2
aire = (p * (p - a) * (p - b) * (p - c)) ** 0.5
cote_carre = (a + b + c) / 4
print("Périmètre du terrain, en m :", a + b + c)
print("Aire du terrain triangulaire, en m² :", round(aire, 1))
print("Aire du carré de même périmètre, en m² :", round(cote_carre * cote_carre, 2))
print("Le triangle est-il plus grand ?", aire > cote_carre * cote_carre)
```

Le programme affiche 158, puis 974.1, puis 1560.25, et enfin `False`.

Le demi-périmètre `p` vaut 79.0. Lui donner un nom n'est pas un confort d'écriture : c'est ce qui rend la ligne de Héron lisible, et c'est ce qui garantit que les quatre `p` de la formule parlent bien du même nombre. Le côté du carré, lui, se déduit du périmètre : quatre côtés égaux pour 158 m de clôture, donc 39,5 m chacun.

La réponse du cahier est le vrai résultat de l'exercice. Avec la même clôture de 158 m, le triangle ne rend que 974,1 m² là où le carré en rend 1 560,25 : 586,15 m² partent en fumée, soit **37,6 % de terrain en moins**.
