## Ce que tu sauras faire

- lire une affectation comme un rangement de valeur, et non comme une égalité mathématique
- passer de la notation ← du langage naturel au signe = de Python, et inversement
- nommer le type d'une variable parmi entier, flottant, chaîne de caractères et booléen
- écrire une formule qui combine plusieurs variables, et la placer au bon endroit de la séquence
- prévoir ce que devient un résultat déjà calculé quand la variable qui l'a produit change

Dans ce carnet, certains programmes sont là pour être **lus** : ils n'ont pas de crochets `[ ]` à leur gauche et ne s'exécutent pas. Les autres sont des cellules de code : clique dedans, puis `Maj` + `Entrée`.

## Cours

À lire avant de commencer.

Un **programme** est une suite d'instructions. Elles sont exécutées **dans l'ordre**, de haut en bas, une seule fois chacune. Échanger deux lignes, c'est changer le programme.

Une **variable** est un nom qui retient une valeur. On la crée par une **affectation**. En langage naturel, l'affectation s'écrit avec une flèche :

    cote ← 5

qui se lit « `cote` reçoit 5 ». En Python, cette flèche s'écrit avec le signe égal.

````python
cote = 5
print(cote)
````

**Ce signe n'est pas une égalité mathématique.** Il ne compare rien : il range une valeur dans une variable. La ligne se lit de droite à gauche — on calcule ce qui est à droite, on le range dans le nom qui est à gauche. C'est pourquoi la ligne `cote = cote + 3` n'a aucun sens en mathématiques, alors qu'elle en a un très précis en Python : prendre la valeur actuelle de `cote`, lui ajouter 3, ranger le total dans `cote`.

**Le piège de cette notion.** Une variable garde la valeur qu'elle avait **au moment où sa ligne a été lue**. Si elle change plus tard, ce qui a déjà été calculé à partir d'elle ne change pas pour autant. Ce piège se voit parfois du premier coup d'œil ; il sait aussi rester totalement silencieux. Tu le rencontreras sous ses deux visages.

Toute valeur a un **type**. Il y en a quatre à connaître.

| Type | Nom en Python | Exemples |
|---|---|---|
| entier | `int` | `4`, `-17`, `0` |
| flottant, c'est-à-dire nombre à virgule | `float` | `3.5`, `2.0` |
| chaîne de caractères | `str` | `"carré"`, `"7"` |
| booléen | `bool` | `True`, `False` |

Trois faits à retenir. La division `/` donne **toujours** un flottant, même quand le quotient tombe juste. Deux chaînes s'ajoutent en se collant l'une à l'autre : `"7" + "7"` donne `"77"`. Une comparaison comme `cote > 10` ou `cote == 5` ne donne pas un nombre mais un booléen — attention : `==` compare, tandis que `=` range. L'instruction `type(cote)` affiche le type d'une variable.

````python
print(7 / 2)
print(type(7 / 2))
````

**La tortue.** Quelques instructions suffisent à dessiner. `forward(60)` avance de 60 pas en laissant un trait. `left(90)` et `right(90)` font pivoter la tortue sur place d'un quart de tour, vers sa gauche ou vers sa droite, sans rien tracer. `reset()` repart d'un écran vide et `done()` termine le dessin. La tortue démarre au centre, tournée vers la droite.
