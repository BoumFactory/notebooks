## Cours

À lire avant de commencer.

Une boucle **bornée** répète des instructions un nombre de fois **connu à l'avance**. On l'écrit avec `for` et `range`. Comme pour un `if`, les deux-points ferment la ligne et l'indentation délimite ce qui est répété.

`range` fabrique une suite d'entiers. La borne de droite n'est **jamais atteinte** : c'est la première valeur que la boucle ne prend pas.

| Ce qu'on écrit | Valeurs prises par `k` | Nombre de tours |
|---|---|---|
| `range(4)` | 0, 1, 2, 3 | 4 |
| `range(1, 5)` | 1, 2, 3, 4 | 4 |
| `range(6, 31, 6)` | 6, 12, 18, 24, 30 | 5 |

Le troisième nombre est le **pas** : la boucle avance alors de 6 en 6.

**Le piège de cette notion : le tour manquant.** Pour aller jusqu'à `b` compris, il faut écrire `range(1, b + 1)`. Cette faute est parfois bruyante — le résultat change à vue d'œil — et parfois **muette** : selon les nombres, elle ne change rien du tout. Tu la rencontreras sous ses deux formes.

Une boucle ne sert pas seulement à répéter : elle fabrique un résultat dans une variable. Trois rôles reviennent sans cesse.

| Rôle | Avant la boucle | Dans la boucle | Ce qu'on obtient |
|---|---|---|---|
| accumulateur | `total = 0` | `total = total + m` | une somme |
| compteur | `compteur = 0` | `compteur = compteur + 1` | un effectif |
| mémoire | `dernier = 0` | `dernier = m` | la dernière valeur retenue |

La variable est créée **avant** la boucle, mise à jour **dans** la boucle, utilisée **après**.

**Dérouler au lieu de deviner.** Le tableau de trace donne, tour par tour, la valeur de chaque variable. C'est le seul outil qui reste disponible en devoir, sans machine.

```python
total = 0
for k in range(1, 4):
    total = total + 10 * k
print(total)
```

| tour | `k` | `total` avant | `total` après |
|---|---|---|---|
| 1 | 1 | 0 | 10 |
| 2 | 2 | 10 | 30 |
| 3 | 3 | 30 | 60 |

Un outil d'arithmétique, enfin. `a % b` est le **reste** de la division euclidienne de `a` par `b`. Donc `a % b == 0` se lit « `a` est un multiple de `b` ».

````python
print(30 % 6)
print(31 % 6)
````

La tortue de la première séance ne change pas : `forward`, `left` et `right`, encadrées par `reset()` et `done()`. Elle sert ici de véhicule à la répétition — un tour de boucle, un morceau de figure.

Revenir à sa direction de départ ne suffit donc pas à refermer une figure. Ce qui décide, c'est le cumul des rotations — nul pour l'escalier, 360° pour le carré.

### ◆ Exercice e1 — ÉCRIRE

Difficulté ●●●○○

Un polygone régulier a tous ses côtés de même longueur et tous ses angles égaux. Pour le tracer, la tortue répète toujours la même chose : avancer d'un côté, puis pivoter du même angle. Quand la figure se referme, la tortue a fait exactement un tour complet.

Écris un programme qui trace un polygone régulier à `n` côtés dont le périmètre vaut 720 pas. Le nombre `n` est posé en tête, dans la cellule ci-dessous.

Deux exigences :

- ton programme doit rester juste si l'on change **uniquement** la ligne `n = 6` ;
- ni la longueur d'un côté ni l'angle ne peuvent être écrits en dur : ils se calculent.

````python
from turtle import *
reset()

n = 6

done()
````

<details>
<summary><strong>Un coup de pouce</strong></summary>

La tortue pivote `n` fois et revient exactement à sa direction de départ : que vaut chaque pivotement ?

Le périmètre est la somme des `n` côtés, tous égaux. Que mesure alors un côté ?

Tes deux réponses sont des divisions. Range chacune dans une variable avant la boucle, plutôt que de la recopier à l'intérieur.

</details>

<details>
<summary><strong>Correction</strong></summary>

```python
from turtle import *
reset()

n = 6
cote = 720 / n
angle = 360 / n
for k in range(n):
    forward(cote)
    left(angle)

done()
```

Les deux mesures se déduisent de `n`, et c'est ce qui rend le programme réutilisable : avec `n = 6`, un côté de 120 et un angle de 60 donnent un hexagone ; avec `n = 9`, un côté de 80 et un angle de 40 donnent un ennéagone ; le périmètre, lui, vaut toujours 720.

Écrire `left(60)` en dur aurait marché — pour l'hexagone, et pour lui seul. C'est la faute que la deuxième exigence interdit : un nombre juste une fois n'est pas un nombre juste.

Le nombre de tours est `n` lui-même : `range(n)` fournit `n` valeurs, de 0 à `n - 1`. Ces valeurs ne servent à rien dans le corps de la boucle, et c'est normal : ici, `k` compte les tours, il ne calcule pas.

Enfin, `720 / n` vaut 120.0 et non 120 : la division `/` donne toujours un flottant. La tortue avance de la bonne longueur sans s'en émouvoir.

</details>

Le **programme 2**, ligne 3, porte une **erreur de syntaxe** : il manque les deux-points à la fin de la ligne `for`. Python n'exécute aucune ligne et signale `SyntaxError: expected ':'`. La faute est bruyante, donc facile à réparer.

<summary><strong>Un coup de pouce</strong></summary>

Deux tests doivent réussir au même tour, pour le même entier : `and` sert exactement à cela, et tu l'as rencontré à la séance sur les conditions.

Il te faut par ailleurs deux variables aux rôles différents : un compteur, et une mémoire qui retient la dernière valeur retenue. Le tableau des trois rôles, dans le cours, dit comment on initialise chacune.

<details>
<summary><strong>Correction</strong></summary>

```python
compteur = 0
plus_grand = 0
for m in range(1, 501):
    if m % 4 == 0 and m % 6 == 0:
        compteur = compteur + 1
        plus_grand = m
print(compteur)
print(plus_grand)
```

Les deux conditions sont réunies par `and` : le tour ne compte que si les deux tests réussissent en même temps. Le programme affiche 41, puis 492.

Ces 41 nombres sont espacés de 12, et 492 vaut 12 × 41. Un entier multiple de 4 et de 6 est exactement un multiple de 12 : c'est ce que ton programme vient d'établir. Écrire `range(12, 501, 12)` aurait donné le bon résultat, mais en supposant connu ce qu'il s'agissait justement de vérifier.

Avec `or` à la place de `and`, le compte serait tout autre : il retiendrait tous les multiples de 4 et tous les multiples de 6, y compris ceux qui ne sont multiples que de l'un des deux.

Voici une figure tracée par la tortue, dessinée à l'échelle. Tous ses segments ont la même longueur, et la tortue est revenue à son point de départ.

Compte les segments. Ils sont tous de même longueur : chaque tour de ta boucle fera donc exactement la même chose, et le seul nombre à trouver est l'angle de pivotement.

Prends un crayon et suis la tortue de segment en segment, en notant chaque fois de combien elle a tourné. Au moment où elle retrouve sa direction de départ, combien de tours complets a-t-elle réellement décrits ? Ce n'est pas un.

</details>

<details>
<summary><strong>Correction</strong></summary>

```python
from turtle import *
reset()

for k in range(5):
    forward(200)
    right(144)

done()
```

Cinq segments égaux, donc cinq tours et un unique pivotement à trouver.

Pour un pentagone ordinaire, la tortue tourne cinq fois de 72°, ce qui fait 360° : un tour complet. Ici, elle passe d'une pointe à la pointe **opposée**, et il lui faut **deux** tours complets pour revenir à sa direction de départ. Les cinq pivotements totalisent donc 720°, et chacun vaut 720 ÷ 5, c'est-à-dire 144°.

Le sens n'est pas indifférent : avec `left(144)`, on obtient la même étoile retournée. C'est le cumul des rotations, et lui seul, qui décide de la forme — un tour ferme un polygone, deux tours ferment une étoile.

### ◆ Exercice a2 — ÉCRIRE

Difficulté ●●●●●

Voici une rosace, dessinée à l'échelle. Tous les carrés sont identiques et partagent un sommet, qui est le point de départ de la tortue.



Retrouve le programme qui la produit.

Pour ce défi, tu auras besoin d'une boucle placée **à l'intérieur** d'une autre. Le principe est celui de l'indentation : ce qui est décalé une fois appartient à la boucle extérieure, ce qui est décalé deux fois appartient à la boucle intérieure et se répète donc à chaque tour de celle-ci.

<details>
<summary><strong>Un coup de pouce</strong></summary>

Commence par le motif : écris seulement le programme qui trace **un** carré. Combien de tours fait cette boucle-là, et de combien la tortue pivote-t-elle à chaque tour ?

Une fois le carré tracé, où se trouve la tortue, et dans quelle direction regarde-t-elle ? C'est ce qui rend possible de recommencer sans se déplacer.

Compte enfin les carrés de la rosace. Entre le premier et le dernier, la tortue a distribué un tour complet : de combien pivote-t-elle entre deux carrés ?

</details>

<details>
<summary><strong>Correction</strong></summary>

```python
from turtle import *
reset()

for tour in range(6):
    for cote in range(4):
        forward(80)
        left(90)
    left(60)

done()
```

La boucle intérieure trace un carré : quatre côtés, quatre quarts de tour. À la fin de ses quatre tours, la tortue est revenue à son point de départ **et** à sa direction de départ — c'est ce qui permet de recommencer sans jamais lever le crayon ni se déplacer.

La boucle extérieure répète ce motif six fois. Entre deux carrés, la ligne `left(60)` fait pivoter la tortue : six pivotements de 60° totalisent 360°, les six carrés se répartissent donc régulièrement autour du sommet commun. Vingt-quatre segments sont tracés en tout.

Avec `left(90)` à la place de `left(60)`, la tortue aurait déjà décrit un tour complet au bout de quatre carrés : les cinquième et sixième se superposeraient exactement aux deux premiers, et la rosace n'aurait plus que quatre branches. Le nombre de motifs et l'angle qui les sépare ne se choisissent pas indépendamment.
